```python
order_amount=int(input('Enter the amount:'))
if (order_amount>=2000):
    discount=15
elif(1500<=order_amount<2000):
    discount=10
elif(1000<=order_amount<1500):
    discount=7
elif(order_amount<1000):
    discount=0
else:
    print('Invalid order amount')
final_amount=order_amount-(order_amount*discount/100)
tax=final_amount*5/100
final_total=final_amount+tax
print('Sub Total=',final_amount)
print('Tax=',tax)
print('Final Total=',final_total)
```
