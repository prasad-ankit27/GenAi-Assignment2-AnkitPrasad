```python
orders=[]
while True:
    print('1-Add order amount to the list.\n')
    print('2-Show all orders and totals after applying discounts.\n')
    print('q-Quit\n')
    choice=input('Enter your choice:')
    if(choice=='1'):
        amount=int(input('Enter order amount:'))
        orders.append(amount)
    elif(choice=='2'):
        for amount in orders:
            if amount>=2000:
                discount=15
            elif 1500<=amount<2000:
                discount=10
            elif 1000<=amount<1500:
                discount=7
            elif amount<1000:
                discount=0
            final_amount=amount-(amount*discount/100)
            print('Order amount=',amount,'->','Discount=',discount,'->','Final Amount=',final_amount)
    elif(choice=='q'or choice=='Q'):
        break
    else:
        print('Invalid input')
        continue

```
